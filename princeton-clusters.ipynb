# Starting on Princeton Clusters

## HPC and You

If you are coming from a wet-lab focused background, getting started on a high performance computing (HPC) system can seem daunting. There are new acronyms and ways of thinking about computers that you may not be used to. Transitioning from a graphical user interface (GUI) experience (e.g., Safari, ImageJ) to interacting with a terminal can feel archaic and strange. The good news is you are not alone! Almost everyone has similar feelings when they get started.

Imagine walking into a lab for the first time and being asked to run a PCR. Maybe you had this experience as an undergrad. First, what does PCR stand for? What reagents do you need and where are they? What volume of master mix do you need to make? How do you use a pipette?! After working on similar and even more complex protocols, it feels like you could run a PCR in your sleep. So as you get started, remember it's not especially difficult but it is very new. Ask questions, be patient with yourself, and know it gets much easier.

For some final advice, know that some people spend their entire days working on HPC systems. These people hate doing things inefficiently, so if something feels clunky or slow to you, someone else has probably come up with a way to make it smoother. Of course, knowing about a specific tool or technique doesn't mean you can use it immediately. Weigh the cost of learning a new tool with the benefits you expect to reap, and make time to learn new things.

### Why move to the cluster?

Speaking of costs and benefits, let's discuss some of the reasons you would want to move your image analyses to the cluster.

1. **Your data is getting too big for your laptop.**
While toy examples are nice for learning, eventually you will acquire large datasets, such as day-long movies or entire well images. Transferring these from the shared drive can take a day, and opening them in ImageJ might overwhelm your laptop's memory. The cluster we will use has around 100 cores and 1TB of RAM, making it suitable for handling large datasets. Additionally, if you have 1000 replicates of an experiment, the cluster can help you analyze them efficiently.
2. **You need a special method/dependency.**
You might need the latest computer vision method to find cells in an image, requiring specific software like PyTorch. Installing these dependencies on your laptop can be cumbersome and time-consuming. Clusters are designed for writing and compiling code, so many dependencies are already installed, simplifying the process.
3. **You are traveling.**
Instead of moving images locally for analysis, you can keep everything on the cluster and connect remotely. This avoids slow transfers on unstable connections and ensures your changes are stored remotely. Additionally, your laptop's battery life will improve since you aren't processing locally.

It is certainly possible to do all your work on the clusters, and once you get started, it's easier than you think. There are downsides, especially with more interactive processes like ImageJ, but they are relatively minor. Personally, I prefer to experience some lag when clicking a button rather than deciding halfway through that I need to move my work to the cluster anyway.

### Cluster Glossary

- Memory (RAM): Working storage for a computer system. When you open an image, it moves from the disk to memory so you can view or analyze it. Typically ranges in size from GBs to TBs.
- Disk Space: Longer-term storage for a computer system. When you shut off a computer, anything in RAM is lost, but anything on disk is saved. Think of working on a Word document and the power goes out—any unsaved changes are lost.
- Core: A single, physical part of a CPU. Your laptop probably has 4 or 8 cores capable of running several processes independently and simultaneously. Cluster nodes can have hundreds of cores, each capable of running multiple threads.
- GPU (Graphical Processing Unit): A highly parallel accelerator originally designed for rendering graphics in games, now used for applications like neural networks and image processing. Software must be designed to utilize a GPU.
- Node: A complete computer system consisting of a CPU, possibly a GPU, memory, and disk space. Each node has access to shared network storage and local scratch storage. Nodes can vary and have specific purposes.
- Cluster: A group of nodes networked together.
- Head/Login Node: The node you first "land on" when you log into the cluster. It has a network connection but should not be used for intensive computations. It's shared by everyone logging in.
- Compute Node: Nodes you get by requesting SLURM jobs. They typically don't have internet access and are used for running computational tasks. You usually get part of a node, such as 4 cores and 100 GB of RAM.
- Interactive Node: A mix between compute and login nodes. They have internet access and are more powerful (e.g., 80 cores, 1TB RAM). Suitable for more compute-heavy code but still a shared resource.
- Quota: The amount of space you have on a specific file system. While the cluster has petabytes of space, individual users are limited to a few TB. Be mindful of your usage before requesting more space.

### Data Storage

Della provides several places to store your data, each with specific use cases. They vary in speed, visibility, and whether the data is backed up. Pay attention to these differences when starting out! Listed from fastest to slowest:
- Scratch: `/scratch/$USER/` This is a physical disk on the node you are currently logged into. It is the fastest storage option because it's directly connected. As the name suggests, it is not permanent or backed up. Some facilities may automatically delete data older than a week. Scratch on one node is not visible on any other node, but you can use as much space as you need. Use this for temporary files or intermediates in a workflow that will run on one node.
- GPFS: `/scratch/gpfs/$USER/` This is a distributed file system similar to scratch but with two key differences: it is visible from any node, and data generally persists until you delete it (though you may occasionally be asked to free up space). It is still fast, not backed up, and can be very large with quota increases. Use this for intermediate and temporary files needed over longer periods and across multiple nodes. It's also good for storing software environments, as they can be recreated with proper documentation.
- Home: `/home/$USER`  This is where you start by default when you log into the cluster. It is fairly fast, backed up, and visible across nodes. However, it starts with a small quota of 10 GB. This is a good place to store your configuration files and software. Reasonable quota increases (up to 100 GB) are likely to be granted.
- Projects: `/projects/BRANGWYNNE/` This is the shared lab space. It is backed up, fairly large, visible everywhere, and slow. Eventually, this will be replaced by Tiger Data, so quota increases are met with some resistance. Store raw sequencing data and important final results here.

Dotfiles are files whose names begin with a .. By default, these are hidden when you list a directory, but they can be easily unhidden. This is not a security feature but a way to hide configuration files so they don't clutter the files you care about.

One of these configuration files is called .bashrc, which stores several commands that run each time you log into the cluster. You can use this file to customize your shell or add new functions. Let's add a few helpful functions now: enable the "Show Dotfiles" option and filter for bashrc. Click on the "More Options" button and select "Edit."

```bash
# User specific aliases and functions
# confirm file removal before deletion, use rm -f to force
alias rm='rm -i'

# make new files group readable and writable by default
umask 002

# copy files from lowrie to della
lowrie_cp() {
	SMB_PATH=$1
	LOCAL_PATH=$2
	EXTENSION=${3:-""}

	mkdir -p $2

	smbclient \
		//lowrie.princeton.edu/Brangwynne/ \
		-U princeton\\$USER \
		-D "${SMB_PATH}" \
		-c "lcd \"${LOCAL_PATH}\"; prompt; recurse; mget \"*${EXTENSION}\""
}

# start cellprofiler
cellprofiler() {
    module load anaconda3/2024.6
    conda activate cellprofiler
    command cellprofiler
}

```

It's up to you if you want to include all of these, but I will assume you have them set up. The `rm` alias is very helpful when you are starting on the command line (I still use it, just in case). Setting the `umask` is good for working with others in the group, but it does give everyone in our group the ability to view and edit your files. The `cellprofiler` and `lowrie_cp` commands are handy shortcuts.

#### More on lowrie_cp

The `lowrie_cp` command is designed for copying entire directories to Della. The syntax is:
```
lowrie_cp lowrie_path della_path [extension]
```

- **lowrie_path:** This is relative to the Brangwynne root directory. On my system, it's `/mnt/brangwynne/`, but it may vary on yours. This gets added by the command, so if you want `/mnt/brangwynne/Shared/blah`, just put `Shared/blah`.

- **della_path:** This is relative to your current directory, or you can provide an absolute path like any other Linux command.

- **extension:** This optional argument is used to limit what gets copied over. If you have raw ND2 files along with cropped TIFFs, you can limit the copy to just the ND2 files by adding `nd2` as the third argument. This will also limit the transfer to only the top directory; without it, the command will copy everything recursively.

Finally, if you have spaces in your paths, wrap them with quotes so it's clear they are one argument, e.g., `"Shared/crazy path"`.

#### Conda and installing cellprofiler, jupyter

Conda is software for setting up independent Python environments. Let's say some software requires Python 3.8 while another requires Python 3.9. Without Conda, you wouldn't be able to use both of these on the same system. Conda also takes care of (most) dependencies.

A similar program is pip, which is used to install Python packages but doesn't handle environment creation. Typically, you can either pip or conda install a package. Given the choice, you should try to conda install everything you can and only use pip at the end. Usually, you will have a command to copy and paste into the terminal to do the installation, then you just `conda activate` whichever environment you want to run.

So, let's open a terminal by clicking on the small black box on the top left of the desktop. This will be a terminal on whichever vis node you selected, running through the desktop. Next, copy and paste this in:

```bash
module load anaconda3/2024.6
conda create -c conda-forge -n my_jupyter jupyterlab matplotlib -y
conda activate my_jupyter
pip install aicsimageio[nd2]
# for czi 
# pip install aicspylibczi>=3.1.1 fsspec>=2022.8.0
```
The first line creates a conda environment named `my_jupyter` and installs a few dependencies. To open ND2 files, we also need something from pip. To make sure we are pip installing into the environment, we activate `my_jupyter` first, then pip install. If you find out later you need something else, you activate the environment and conda or pip install what you need. If you also work with CZI files, you can run the last, commented out pip install line. Agree to any licenses and installation requests as needed.  This will take a few minutes to complete.

Again for CellProfiler:
```bash
conda create -c conda-forge -n cellprofiler python=3.8 mysqlclient=1.4.6 numpy=1.24.4 python-javabridge wxPython=4.2.0 -y
conda activate cellprofiler
pip install cellprofiler==4.2.6
```

You only have to do that once. In the next steps (and in the future), you only have to use these environments.
